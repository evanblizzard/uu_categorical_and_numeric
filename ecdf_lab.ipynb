```python
import pandas as pd
import numpy as np
import random
import seaborn as sns
import matplotlib as plt
```

# Q1

```python
k = 50000
grid = np.linspace(1/k, 1, endpoint=True, num=k)
grid
```

```python
n = 5_000
random_list = []
for k in range(n):
    random_list.append(random.choice(grid))
    
```

```python
#random_list
```

```python
sns.ecdfplot(random_list)
```

The distribution generated approaches the uniform distribution function as K increases.

# Q5

```python
n = 5000
K = 20

data = np.random.uniform(0, 1, size=(n,K))
sorted_data = np.sort(data, axis=1)

for r in range(1, K+1):
    rank_data = sorted_data[:, r-1] # column for this rank. np arrays allow 2d indexing GOOD TO KNOW!
    sns.ecdfplot(rank_data)
```

This generates variations on the Beta distribution with varying betas and alphas.
