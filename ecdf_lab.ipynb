# Q5

```python
n = 5000
K = 20

data = np.random.uniform(0, 1, size=(n,K))
sorted_data = np.sort(data, axis=1)

for r in range(1, K+1):
    rank_data = sorted_data[:, r-1] # column for this rank. np arrays allow 2d indexing GOOD TO KNOW!
    sns.ecdfplot(rank_data)
```

This generates variations on the Beta distribution with varying rank (beta) and draws (alpha). Higher ranks mean bigger alphas with a mass shift towards 1. Lower ranks mean bigger betas with shifts towards 0. 

# Q7

```python
K = 100_000
n = 5_000
s = 50
ranks = np.random.randint(1, K+1, size=n)
sizes = (K / ranks) ** s

sns.ecdfplot(sizes)
plt.xscale('log') # looked like a vertical line without log scale
```

As K becomes large, this generates a Pareto distribution with the shape parameter b = 1/s. As s increases, the shape parameter b decreases, giving the distribution a heavier tail. That means greater inequality between largest and smallest sizes. As s decreases, the distribution flattens out and sizes across ranks become more similar.
